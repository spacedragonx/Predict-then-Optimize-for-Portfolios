```python
import pandas as pd
import numpy as np
```

```python
file_path = "data_csv/25yr_daily_data_with_spy.csv"
df = pd.read_csv(
    file_path,
    index_col=['symbol', 'timestamp'],  # Restore MultiIndex
    parse_dates=['timestamp']           # Make timestamp datetime
)
print(df.head())
```

```python
print(df.info())
```

```python
close_data = df['close'].unstack(level=0)
close_data.index = pd.to_datetime(close_data.index)
```

```python
print(close_data.head())
```

```python
print(close_data.index)
```

```python
close_data.index = pd.to_datetime(close_data.index).tz_localize(None).normalize()
print(close_data.index[0]) 
```

```python
print(close_data.info())
```

```python
print(close_data.index[1]) 
```

```python
# Log returns

returns = np.log(close_data/close_data.shift(1)).dropna()
```

```python
spy_return = returns['SPY']
assets_ret = returns.drop(columns=['SPY'])
```

```python
features = pd.DataFrame(index = assets_ret.index)
```

```python
feature_list = []
```

```python
for ticker in assets_ret.columns:
    temp_df = pd.DataFrame(index=assets_ret.index)

    # Lagged returns

    temp_df[f'{ticker}_ret_lag1'] = assets_ret[ticker]
    temp_df[f'{ticker}_ret_lag2'] = assets_ret[ticker].shift(1)
    temp_df[f'{ticker}_ret_lag3'] = assets_ret[ticker].shift(2)

    # VOlatility

    temp_df[f'{ticker}_vol_21'] = assets_ret[ticker].rolling(window=21).std()

    # Corelation with the market index (S&P 500)

    temp_df[f'{ticker}_corr_spy'] = assets_ret[ticker].rolling(window=21).corr(spy_return)

    # Relative returns to market index (S&P 500)
    temp_df[f'{ticker}_rel_spy'] = assets_ret[ticker] - spy_return

    feature_list.append(temp_df)

all_features = pd.concat(feature_list,axis=1).dropna()

print(all_features.head())
```

```python
targets = assets_ret.shift(-1).loc[all_features.index]
```

```python
valid_indices = targets.dropna().index
X = all_features.loc[valid_indices]
y = targets.loc[valid_indices]

print(f"Features Shape: {X.shape}") # (Time, Num_Stocks * Num_Features)
print(f"Targets Shape:  {y.shape}") # (Time, Num_Stocks)
```

```python
print(X.info())
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
import torch

data_package = {
    'X': torch.tensor(X_scaled, dtype=torch.float32),   # Inputs
    'y': torch.tensor(y.values, dtype=torch.float32),   # Targets
    'dates': X.index.astype(str).tolist(),                          # Dates for plotting/backtests
    'tickers': assets_ret.columns.tolist()              # Asset names
}

torch.save(data_package, 'data_save/25_spo_data.pt')
```
